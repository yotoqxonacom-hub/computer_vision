```python
import cv2
```

```python
from google.colab.patches import cv2_imshow

image = cv2.imread('/cat.jpeg')

cv2_imshow(image)
```

```python
print("SHAPE:", image.shape)
print("HEIGHT:", image.shape[0])
print("WIDTH:", image.shape[1])
print("CHANNELS:", image.shape[2])
```

```python
(b, g, r) = image[100, 50]
print('blue:', b, "Green:", g, "Red:", r)
```

```python
gray = cv2.cvtColor(image, cv2.COLOR_BGR2GRAY)
cv2_imshow(gray)
```

```python
resized = cv2.resize(gray,(300,300))
cv2_imshow(resized)
```

```python
cropped = gray[100:300, 150:350]
cv2_imshow(cropped)
```

```python
rotated = cv2.rotate(gray, cv2.ROTATE_90_CLOCKWISE)
cv2_imshow(rotated)
```

```python
flipped = cv2.flip(gray, 1)
cv2_imshow(flipped)
```

```python
cv2.imwrite('cropped_cat.jpg', cropped)
```

opencv_video streaming **codes**

```python
import cv2
print('HELLO WORLD')

cap = cv2.VideoCapture(0)

face_cascade = cv2.CascadeClassifier(
    cv2.data.haarcascades + "haarcascade_frontalface_default.xml"
)

while True:
    success, frame = cap.read()

    if not success:
        break

    gray = cv2.cvtColor(frame, cv2.COLOR_BGR2GRAY)

    faces = face_cascade.detectMultiScale(gray, 1.1, 4)

    for (x, y, w, h) in faces:
        cv2.rectangle(frame, (x, y), (x + w, y + h), (0, 255, 0), 2)

    cv2.imshow("Video", frame)

    if cv2.waitKey(1) & 0xFF == ord('q'):
        break

cap.release()
cv2.destroyAllWindows()


# <====> #

# cap = cv2.VideoCapture(0)

# while True:
    # success, frame = cap.read()  # boolean
    # if not success:
    #   break

    # gray = cv2.cvtColor(frame, cv2.COLOR_BGR2GRAY)
    # blur = cv2.GaussianBlur(gray, (15, 15), 0)
    # edges1 = cv2.Canny(frame, 100, 200)
    # edges2 = cv2.Canny(frame, 50, 250)
    # flipped_frame = cv2.flip(frame, 1)  # flip the frame horizontally

    # cv2.imshow('Video', frame)
    # cv2.imshow('Video1', gray)
    # cv2.imshow('Video2', blur)  # for noise reduction
    # cv2.imshow('Video4', edges2)  # for edge detection
    # cv2.imshow('Video5', flipped_frame)  # for flipped frame

    # if cv2.waitKey(1) & 0xFF == ord('q'):
    # break


# cap.release()
# cv2.destroyAllWindows()
```
